```python
# TASK 10 - AI vs ML vs DEEP LEARNING
# AI & ML Internship - Veda Technology

comparison = [
    {
        "Concept": "Artificial Intelligence (AI)",
        "Definition": "A broad field of creating systems that can perform tasks requiring human-like intelligence.",
        "Scope": "Broadest",
        "Examples": "Virtual assistants, recommendation systems, game-playing systems"
    },
    {
        "Concept": "Machine Learning (ML)",
        "Definition": "A subset of AI in which systems learn patterns from data and improve their performance.",
        "Scope": "Subset of AI",
        "Examples": "Spam detection, recommendation systems, fraud detection"
    },
    {
        "Concept": "Deep Learning (DL)",
        "Definition": "A subset of ML that uses multi-layer neural networks to learn complex patterns.",
        "Scope": "Subset of ML",
        "Examples": "Image recognition, speech recognition, autonomous driving"
    }
]

print("AI vs ML vs DEEP LEARNING")
print("=" * 80)

for item in comparison:
    print("\nConcept:", item["Concept"])
    print("Definition:", item["Definition"])
    print("Scope:", item["Scope"])
    print("Examples:", item["Examples"])

print("\nRelationship:")
print("AI is the broad field.")
print("ML is a subset of AI.")
print("Deep Learning is a subset of ML.")

print("\nSimple relationship:")
print("AI")
print(" └── Machine Learning")
print("      └── Deep Learning")
```
